# Task 21: Missing Value Analysis

## Objective
Understand the importance of identifying missing data before feeding it into machine learning models. Missing data can lead to biased models, convergence failures in optimization algorithms, or runtime errors.

## Deliverables
1. Missing-value summary dataframe (Total counts & Percentages).
2. Column-wise missing-value counts.
3. Identification of columns requiring treatment.

```python
import pandas as pd

# Load the Titanic dataset (ensure titanic.csv is in the same directory)
df = pd.read_csv('titanic.csv')

print("--- Dataset Overview ---")
print(f"Total Rows: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}\n")

# 1. Column-wise missing value counts
missing_counts = df.isnull().sum()

# 2. Calculate percentage of missing values
missing_percentage = (df.isnull().mean() * 100).round(2)

# 3. Combine into a summary DataFrame
missing_summary = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage (%)': missing_percentage
})

# Filter out columns with 0 missing values to focus on what needs treatment
missing_summary_filtered = missing_summary[missing_summary['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)

print("--- Missing Values Summary ---")
display(missing_summary_filtered)

# Alternatively, display the full table if needed:
# display(missing_summary)
```

## Key Findings & Identification of Columns Requiring Treatment

From the analysis of the Titanic dataset, the following columns contain missing values and require proper imputation or removal strategies before machine learning modeling:

* **Cabin**: Has a massive percentage of missing values (usually > 75%). Often dropped entirely or converted into a binary feature like `Has_Cabin` (1 or 0).
* **Age**: Contains several missing numerical entries. Typically requires imputation using the median or mean (or advanced methods like KNN imputation / group-wise median based on `Pclass` and `Sex`).
* **Embarked / Embarked_Town**: Contains a couple of missing categorical rows. Can be filled using the mode (most frequent value).

# Interview Questions & Answers

### 1. What is a missing value?
* **Answer:** A missing value refers to data that is absent or not stored for a specific variable (feature) in an observation. In Pandas/Python, these are typically represented as `NaN` (Not a Number), `None`, or `NaT` (Not a Time).

### 2. Why can missing values be a problem in machine learning?
* **Answer:** 
  1. **Algorithm Incompatibility:** Most standard machine learning algorithms (like scikit-learn implementations of Linear Regression, Support Vector Machines, or Neural Networks) cannot handle `NaN` values natively and will throw a runtime error.
  2. **Model Bias:** If missingness is not random, deleting rows or using naive imputation can introduce severe statistical bias, reducing the model's predictive power and generalizability.
  3. **Loss of Information:** Completely dropping rows with missing values can discard valuable data, especially if multiple features have scattered missing entries.

### 3. How can you identify missing values in Pandas?
* **Answer:** You can identify missing values using the following Pandas functions:
  * `.isnull()` or `.isna()`: Returns a boolean DataFrame/Series indicating `True` where values are missing.
  * `.sum()`: Chained after `.isnull()`, it counts the total number of missing values per column (e.g., `df.isnull().sum()`).
  * `.info()`: Provides a quick summary of non-null counts per column.
  * Heatmaps or visualization libraries like `missingno` for a visual distribution of missing data.
